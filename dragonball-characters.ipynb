```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

filepath = "/kaggle/input/dragon-ball-synthetic-stats-for-ml-practice/dragonball_characters.csv"
data_db = pd.read_csv(filepath)

sns.barplot(x=data_db["Race"].value_counts().index,
            y=data_db["Race"].value_counts().values)

plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x=data_db["Race"])
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(x=data_db["Alignment"])
plt.xticks(rotation=45)
plt.show()
```

```python
avg_power = data_db.groupby("Race")["MaxPowerLevel"].mean().reset_index()

sns.barplot(x="Race", y="MaxPowerLevel", data=avg_power)
plt.xticks(rotation=45)
plt.show()
```

```python
avg_age = data_db.groupby("Saga")["Age"].mean().reset_index()

sns.barplot(x="Saga", y="Age", data=avg_age)
plt.xticks(rotation=45)
plt.show()
```

```python
power_saga = data_db.groupby("Saga")["MaxPowerLevel"].mean().reset_index()

sns.lineplot(x="Saga", y="MaxPowerLevel", data=power_saga, marker="o")
plt.xticks(rotation=45)
plt.show()
```

```python
pivot_age = pd.pivot_table(data_db, values="Age", index="Race", columns="Saga")

sns.heatmap(pivot_age, annot=True, fmt=".0f")
plt.show()
```

```python
sns.pairplot(data_db[["BasePowerLevel", "MaxPowerLevel", "Age", "HasKiControl"]])
plt.show()
```

```python
sns.jointplot(x="BasePowerLevel", y="MaxPowerLevel", data=data_db, kind="kde")
plt.show()
```

```python
sns.violinplot(x="Race", y="MaxPowerLevel", data=data_db)
plt.xticks(rotation=45)
plt.show()
```

```python
sns.swarmplot(x="Race", y="Age", data=data_db)
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(x="Saga", y="MaxPowerLevel", data=data_db)
plt.xticks(rotation=45)
plt.show()
```
